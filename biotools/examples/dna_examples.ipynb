In `BioTools`, sequences are represented as `DNA` objects. There are multiple classes that correspond to different bimolecules\
that we will work with in this tutorial. First, let's make a `DNA` object below. Run the cell below to create an object and\
visualize its representation.

````python
from biotools.dna import DNA

test_dna = DNA(seq="ATGCATGCATGC", name="foo")
test_dna
````

`DNA` is represented in the form (`Name`, `BioMolecule Enum`, `Length`, `Annotations`, `Strandedness`) \
\
`Length` provides you the length in number of base pairs.\
`Annotations` provides you with the number of annotations of the sequence.\
`Strandedness` tells you if the sequence is single-stranded, double-stranded, circular, or cut.\
\
We'll dive into the specifics of these properties and more later. For now, let's take a look at the sequence:

````python
test_dna.sequence()
print()
print("Top Strand: ", test_dna.top_strand)
print("Bottom Strand: ", test_dna.bottom_strand)
````

What we see above is the representation at the sequence level, showing the top and bottom strands.\
If we wanted to visualize each separately, we can check the `top_strand` and `bottom_strand` properties\
of the sequence.\
NOTE: Each will be shown 5' --> 3', so the bottom strand is just the reverse complement of the top.\
\
Being able to generate random sequences is important both for testing and library design. Let's take a look\
at `bio_utils` to see how we can do that:

````python
from biotools.bio_utils import generate_random_sequence
from biotools.bio_enums import BioMolecule

random_dna = generate_random_sequence(25, BioMolecule.DNA)
random_dna.top_strand
````

Above will generate a random sequence. Keep re-running the cell to see it in action!\
If we want to generate a pool of such sequences, we can leverage `n_seqs` in the `generate_random_sequence` function.

````python
random_dna_list = generate_random_sequence(25, BioMolecule.DNA, 10)
random_dna_list
````

We can also pull sequences from common file types! Supported files are:
1. Genbank (.gb/.gbk)
2. FASTA (.fa/.fas/.fasta)

Let's see how we can work with these:

````python
from biotools.bio_io import BioFileParser
import pprint

# FASTA FILES #

# BioFileParser will load the sequence present in the FASTA file
fasta_seq = BioFileParser().parse_fasta("../test_seqs/test2.fasta")
print(fasta_seq)
print()

# We can also pull multiple sequences from a FASTA file if there are more than one!
multi_fasta_seq = BioFileParser().parse_fasta("../test_seqs/TEST.fasta")
pprint.pprint(multi_fasta_seq)
print()

# GENBANK FILES #
genbank_seq = BioFileParser().parse_genbank("../test_seqs/y4_onestep_vector.gb")
print(genbank_seq)
print()

# NOTE: BioFileParser assumes that FASTA sequences are linear unless specified via 
# the `circular` argument
circular_fasta_seq = BioFileParser().parse_fasta("../test_seqs/test2.fasta", circular=True)
print(circular_fasta_seq)
print()
````

If we wanted to generate a primer from a DNA sequence, we can easily do this using `to_oligo`!\
If we want the oligo to be a section of the DNA sequence, we can use the `slice` argument.

````python
test_oligo = test_dna.to_oligo()
print(test_oligo)
print(test_oligo.top_strand)
print()

test_oligo_slice = test_dna.to_oligo(slice=(0,5))
print(test_oligo_slice)
print(test_oligo_slice.top_strand)
````
